```python
%pip install openai langchain-community
import os
from openai import OpenAI
from langchain_community.chat_message_histories import ChatMessageHistory

# Setup the Local Client
client = OpenAI(
    base_url="http://localhost:1234/v1", 
    api_key="lm-studio"
)

history = ChatMessageHistory()

print("Chatbot initialized for Charles Arvin N. Torres.")
print("Type 'exit' anytime to stop.\n")

while True:
    query = input("User Input: ") 

    if query.lower() == "exit":
        print("\nExiting...")
        print("Program Finished.")
        break

    if not query.strip():
        continue

    # Add the current user query to the history object
    history.add_user_message(query)

   
    
    print(f"--- Chat History for Charles Arvin N. Torres ---")
    
    # We loop through the stored history to display it on screen
    for msg in history.messages:
        role_label = "User" if msg.type == "human" else "Assistant"
        print(f"{role_label}: {msg.content}")
        if msg.type == "human":
            print("- " * 30)

    # Convert history to the list format the API expects
    messages_for_api = [
        {"role": "user" if m.type == "human" else "assistant", "content": m.content}
        for m in history.messages
    ]

    try:
        # Call the model
        stream = client.chat.completions.create(
            model="qwen3-1.7b", 
            messages=messages_for_api,
            stream=True
        )

        print("Assistant: ", end="", flush=True)
        full_ai_response = ""

        for chunk in stream:
            if chunk.choices[0].delta.content:
                content = chunk.choices[0].delta.content
                print(content, end="", flush=True)
                full_ai_response += content

        # Save AI response so it's remembered next turn
        history.add_ai_message(full_ai_response)

        print("\n\n" + "=" * 100 + "\n")

    except Exception as e:
        print(f"\n❌ Error: {e}")
        print("=" * 100 + "\n")
```
